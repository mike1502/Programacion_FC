## 1.1 Conceptos Generales

La **Programación Orientada a Objetos (POO)** es un paradigma de programación que organiza el diseño de software alrededor de **datos** (u objetos), en lugar de funciones y lógica aislada.

### Los 4 Pilares de la POO
1. **Abstracción:** Simplificar un objeto del mundo real enfocándose solo en sus características relevantes para el problema (ej. de un *Estudiante* nos interesa su matrícula y promedio, no su color favorito).
2. **Encapsulamiento:** Agrupar datos (atributos) y comportamientos (métodos) dentro de una misma unidad (clase), restringiendo el acceso directo a ciertos componentes para proteger la integridad de los datos.
3. **Herencia:** Mecanismo mediante el cual una clase nueva adquiere las propiedades y comportamientos de una clase existente, fomentando la reutilización de código.
4. **Polimorfismo:** Capacidad de diferentes clases para responder de manera distinta a un mismo mensaje o método (ej. el método `hablar()` produce `"Guau"` en un Perro y `"Miau"` en un Gato).

### Paradigma Estructurado vs. Paradigma Orientado a Objetos
En el enfoque estructurado, los datos y las funciones están separados. En la POO, los datos y las funciones que los manipulan viven juntos.

````python
# COMPARATIVA PRÁCTICA: Enfoque Estructurado vs. Enfoque POO

# --- 1. ENFOQUE ESTRUCTURADO (Datos y funciones separados) ---
alumno_nombre = "Ana López"
alumno_calificaciones = [9, 8, 10]

def calcular_promedio_estructurado(calificaciones):
    return sum(calificaciones) / len(calificaciones)

print(f"[Estructurado] Promedio de {alumno_nombre}: {calcular_promedio_estructurado(alumno_calificaciones):.2f}")


# --- 2. ENFOQUE ORIENTADO A OBJETOS (Datos y comportamiento unidos) ---
class Estudiante:
    def __init__(self, nombre, calificaciones):
        self.nombre = nombre
        self.calificaciones = calificaciones

    def calcular_promedio(self):
        return sum(self.calificaciones) / len(self.calificaciones)

# Creamos un objeto (instancia)
estudiante_1 = Estudiante("Ana López", [9, 8, 10])
print(f"[POO] Promedio de {estudiante_1.nombre}: {estudiante_1.calcular_promedio():.2f}")
````

## 1.2 Metodología de Diseño

Antes de escribir código en Python, un ingeniero de software diseña la solución. El proceso básico consta de tres pasos:

1. **Identificación de Sustantivos (Clases):** Analizar el enunciado del problema para detectar entidades clave (ej. *CuentaBancaria*, *Cliente*, *Transacción*).
2. **Identificación de Adjetivos/Propiedades (Atributos):** Determinar qué datos definen el estado de cada entidad (ej. *saldo*, *numero_cuenta*, *titular*).
3. **Identificación de Verbos/Acciones (Métodos):** Definir qué operaciones puede realizar o recibir cada entidad (ej. *depositar()*, *retirar()*, *consultar_saldo()*).

### Modelado Visual (Diagramas UML básicos)
En diseño orientado a objetos solemos representar las clases como cajas divididas en tres secciones:
```text
+-----------------------+
|    CuentaBancaria     |  <-- Nombre de la Clase
+-----------------------+
| - titular: str        |  <-- Atributos
| - saldo: float        |
+-----------------------+
| + depositar(monto)    |  <-- Métodos
| + retirar(monto)      |
+-----------------------+

````python
#### [Celda 4: Código]
```python
# TRADUCCIÓN DE UN DISEÑO CONCEPTUAL A CÓDIGO PYTHON

class CuentaBancaria:
    """Representa una cuenta bancaria básica según nuestro diseño."""

    def __init__(self, titular: str, saldo_inicial: float = 0.0):
        # Atributos
        self.titular = titular
        self.saldo = saldo_inicial

    # Métodos (Acciones)
    def depositar(self, monto: float):
        if monto > 0:
            self.saldo += monto
            print(f"Depósito exitoso de ${monto}. Nuevo saldo: ${self.saldo}")
        else:
            print("El monto a depositar debe ser positivo.")

    def retirar(self, monto: float):
        if 0 < monto <= self.saldo:
            self.saldo -= monto
            print(f"Retiro exitoso de ${monto}. Saldo restante: ${self.saldo}")
        else:
            print("Fondos insuficientes o monto inválido.")

# Prueba del diseño
cuenta = CuentaBancaria("Carlos Ruiz", 1000.0)
cuenta.depositar(500.0)
cuenta.retirar(200.0)
````

## 1.3 Organización General de un Programa

En proyectos reales de Python (fuera de un cuaderno de Colab), el código se organiza en **módulos** (archivos `.py`) y **paquetes** (carpetas con un archivo `__init__.py`).

Una estructura profesional típica en GitHub luce así:

```text
proyecto_universidad/
│
├── README.md               # Documentación del proyecto
├── main.py                 # Punto de entrada de la ejecución
└── modelos/                # Paquete con nuestras clases
    ├── __init__.py         # Indica a Python que esta carpeta es un paquete
    ├── estudiante.py       # Contiene la clase Estudiante
    └── curso.py            # Contiene la clase Curso

````python
#### [Celda 6: Código]
```python
# SIMULACIÓN DE ORGANIZACIÓN DE UN PROGRAMA EN PYTHON

# 1. Sección de importaciones (Librerías estándar o módulos propios)
import math

# 2. Definición de Clases y Funciones
class Circulo:
    def __init__(self, radio: float):
        self.radio = radio

    def calcular_area(self) -> float:
        return math.pi * (self.radio ** 2)

# 3. Bloque principal de ejecución
# Este bloque solo se ejecuta si el archivo se corre directamente,
# y NO si es importado desde otro archivo.
if __name__ == "__main__":
    print("--- Ejecutando programa principal ---")
    c = Circulo(5)
    print(f"El área del círculo con radio {c.radio} es: {c.calcular_area():.2f}")
````

## 1.4 Normas de Estilo (PEP 8)

Para que el código sea legible por otros desarrolladores y cumpla con estándares profesionales, Python cuenta con una guía de estilo oficial llamada **PEP 8**.

### Reglas esenciales para POO en Python:
* **Nombres de Clases:** Usar `PascalCase` (ej. `CuentaBancaria`, `RegistroAcademico`).
* **Nombres de Métodos, Funciones y Variables:** Usar `snake_case` (ej. `calcular_promedio`, `saldo_actual`).
* **Constantes:** Usar `MAYUSCULAS_CON_GUION_BAJO` (ej. `TASA_INTERES = 0.05`).
* **Indentación:** Siempre **4 espacios** por nivel (nunca mezclar tabuladores y espacios).
* **Docstrings:** Documentar el propósito de las clases y métodos usando comillas triples `"""..."""`.
* **Longitud de línea:** Máximo 79 caracteres por línea.

````python
# EJEMPLO DE APLICACIÓN DE PEP 8

TASA_IVA = 0.16  # Constante en MAYÚSCULAS


class ProductoInventario:  # Clase en PascalCase
    """Clase que modela un artículo dentro de un almacén."""

    def __init__(self, nombre_producto: str, precio_base: float):  # snake_case
        self.nombre_producto = nombre_producto
        self.precio_base = precio_base

    def obtener_precio_final(self) -> float:  # snake_case y Type Hints
        """Calcula el precio incluyendo el IVA vigente."""
        impuesto = self.precio_base * TASA_IVA
        return self.precio_base + impuesto


# Instanciación siguiendo espacios limpios alrededor de operadores (=)
laptop = ProductoInventario("Laptop Pro 15", 15000.00)
print(f"Producto: {laptop.nombre_producto} | Precio con IVA: ${laptop.obtener_precio_final():,.2f}")
````
